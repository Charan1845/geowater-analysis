```python
dams = pd.read_excel(
    "../Dataset/WDG complete.xlsx",
    sheet_name="Dams & Reservoirs",
    header=1
)

dams.head()
```

```python
# Remove rows where District is empty
dams = dams[dams["District"].notna()]

# Convert Capacity to numeric
dams["Capacity (MCM)"] = pd.to_numeric(
    dams["Capacity (MCM)"],
    errors="coerce"
)

# Remove rows without capacity
dams = dams[dams["Capacity (MCM)"].notna()]
```

```python
district_capacity = (
    dams.groupby("District", as_index=False)["Capacity (MCM)"]
        .sum()
)

district_capacity
```

```python
wdg["District"] = wdg["District"].str.strip().str.title()
district_capacity["District"] = district_capacity["District"].str.strip().str.title()
```

```python
wdg = wdg.merge(
    district_capacity,
    on="District",
    how="left"
)
```
