```python
import pandas as pd

df = pd.read_excel(r"C:\Users\ravip\Desktop\Major Project\Output\WDG_Final.xlsx")

print(df.head())
```

```python
print("Shape:", df.shape)
print("\nDistricts:", df["District"].nunique())
print("\nColumns:")
print(df.columns)
```

```python
# Function to calculate average rainfall only on rainy days
def rainy_day_avg(x):
    rainy = x[x > 0]
    return rainy.mean() if len(rainy) > 0 else 0

district_avg = (
    df.groupby("District")
      .agg(
          Avg_Rain_All_Days=("Rain (mm)", "mean"),
          Avg_Rain_Rainy_Days=("Rain (mm)", rainy_day_avg),
          Rainy_Days=("Rain (mm)", lambda x: (x > 0).sum()),
          Capacity_MCM=("Capacity (MCM)", "max")
      )
      .reset_index()
)
```

```python
district_avg
```

```python
district_avg["Avg_Rain_All_Days"] = district_avg["Avg_Rain_All_Days"].round(2)
district_avg["Avg_Rain_Rainy_Days"] = district_avg["Avg_Rain_Rainy_Days"].round(2)
district_avg["Capacity_MCM"] = district_avg["Capacity_MCM"].round(2)

district_avg
```

```python
district_avg.to_excel("../Output/District_Averages.xlsx", index=False)

print("District_Averages.xlsx created successfully!")
```

```python
df = pd.read_excel("District_Level_Rainfall.xlsx")

# Convert Date
df["Date"] = pd.to_datetime(df["Date"])

# Create Year and Month if they don't exist
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month

while True:

    district = input("Enter District : ").strip()
    year = int(input("Enter Year     : "))

    # Filter district, year and monsoon months
    data = df[
        (df["District"].str.lower() == district.lower()) &
        (df["Year"] == year) &
        (df["Month"].isin([6, 7, 8, 9]))
    ]

    if data.empty:
        print("\nNo data found.")
    else:
        cumulative = data["Rain (mm)"].sum()
        average = data["Rain (mm)"].mean()
        rainy_days = (data["Rain (mm)"] > 0).sum()

        print("\n" + "="*50)
        print("      MONSOON RAINFALL REPORT")
        print("="*50)
        print(f"District                  : {district.title()}")
        print(f"Year                      : {year}")
        print(f"Rainy Days                : {rainy_days}")
        print(f"Cumulative Rainfall       : {cumulative:.2f} mm")
        print(f"Average Daily Rainfall    : {average:.2f} mm/day")
        print("="*50)

    again = input("\nSearch another? (Y/N): ").lower()
    if again != "y":
        break
```
